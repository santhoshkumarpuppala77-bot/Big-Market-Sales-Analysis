```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import OrdinalEncoder,OneHotEncoder,StandardScaler,PowerTransformer,LabelEncoder,RobustScaler
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.neighbors import KNeighborsRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.svm import SVR
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from xgboost import XGBRegressor
from catboost import CatBoostRegressor
from sklearn.model_selection import GridSearchCV,train_test_split,cross_val_score,KFold
from sklearn.metrics import r2_score,root_mean_squared_error,mean_absolute_error,mean_squared_error
```

```python
train4 = pd.read_csv(r"C:\Users\puppa\Documents\Innomatics\Project DS\Big Market Sales Analysis\train_v9rqX0R.csv")
```

```python
train4["Item_Fat_Content"] = (train4["Item_Fat_Content"].replace({"LF": "Low Fat","low fat": "Low Fat","reg": "Regular"}))
```

```python
train4.groupby(train4["Item_Visibility"].eq(0))["Item_Outlet_Sales"].agg(["count", "mean", "median"])
```

```python
train4["Outlet_Age"] = (2013 - train4["Outlet_Establishment_Year"])
```

```python
X4 = train4.drop(columns=["Item_Outlet_Sales",'Item_Identifier','Outlet_Establishment_Year'])
y4 = train4["Item_Outlet_Sales"]
```

```python
X_train,X_test,y_train,y_test = train_test_split(X4,y4,test_size=0.2,random_state=27)
```

```python
cat_nom_cols4 = ["Item_Type","Outlet_Type","Outlet_Identifier"]
num_cols4 = ["Item_Weight","Item_Visibility","Item_MRP"]
cat_ord_cols4 = ['Item_Fat_Content','Outlet_Size','Outlet_Location_Type']
```

```python
cat_order = [['Low Fat', 'Regular'],
             ['Small', 'Medium', 'High'],
             ['Tier 1', 'Tier 2', 'Tier 3']]
```

```python
cat_ordinal4 = Pipeline([('imputer',SimpleImputer(strategy='most_frequent')),("Encoding",OrdinalEncoder(categories=cat_order))])
cat_ordinal4
```

```python
cat_nominal4 = Pipeline([('imputer',SimpleImputer(strategy='most_frequent')),("Encoding",OneHotEncoder(sparse_output=True))])
cat_nominal4
```

```python
num_col_pipeline4 = Pipeline([('Imputation',SimpleImputer(strategy='median'))])#,('Scaling',RobustScaler()),('transformer',PowerTransformer())
num_col_pipeline4  
```

```python
preprocessing4 = ColumnTransformer([('Nominal',cat_nominal4,cat_nom_cols4),
                                  ('Ordinal',cat_ordinal4,cat_ord_cols4),
                                  ('num',num_col_pipeline4,num_cols4)])
preprocessing4
```

```python
CAT = Pipeline([('Preprocessing',preprocessing4),('Catboost',CatBoostRegressor(verbose=0,iterations=502,learning_rate=00.01164955033304544,random_seed=27,l2_leaf_reg=0.005244482153929041,depth = 4))])
CAT.fit(X_train,y_train)
y_pred = CAT.predict(X_test)
print(f' r2 score:', r2_score(y_test,y_pred))
print(f' mean squared error:', mean_squared_error(y_test,y_pred))
print(f' root mean squared error:', root_mean_squared_error(y_test,y_pred))
print(f' mean absolute error:', mean_absolute_error(y_test,y_pred))
```

```python
kf=KFold(n_splits=5,shuffle=True,random_state=27)
cross_val_score(CAT,X_train,y_train,scoring='r2',cv=kf).mean()
```

```python
training_pred  =y_pred = CAT.predict(X_train)

print(f' r2 score:', r2_score(y_train,training_pred))
print(f' mean squared error:', mean_squared_error(y_train,training_pred))
print(f' root mean squared error:', root_mean_squared_error(y_train,training_pred))
print(f' mean absolute error:', mean_absolute_error(y_train,training_pred))
```
